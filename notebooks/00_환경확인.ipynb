# 00. 환경 확인

발제 '프로젝트 시작 전 준비' 항목: 생성 모델 호출 1건과 임베딩 1건을 실행하고 모델명·오류 여부를 확인합니다.

커널은 **AX_RAG (.venv 3.13)** 를 선택하세요.

```python
# 프로젝트는 uv sync로 패키지(rag_home_api)로 설치되므로 바로 불러올 수 있음
from rag_home_api.config import settings

print("생성 모델:", settings.llm_model)
print("재정렬 모델:", settings.rerank_model)
print("임베딩 모델:", settings.embedding_model)
```

```python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings

llm = ChatOpenAI(model=settings.llm_model, api_key=settings.openai_api_key)
reply = llm.invoke("전입신고를 한 문장으로 설명해 줘.")
print(reply.response_metadata.get("model_name"), "→", reply.content)

emb = OpenAIEmbeddings(model=settings.embedding_model, api_key=settings.openai_api_key)
vec = emb.embed_query("보증금 반환")
print("임베딩 차원:", len(vec))
```
